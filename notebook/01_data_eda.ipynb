## Load data

```python
import json
import pandas as pd
from src.config import DATA_DIR

origin_data_path = DATA_DIR / "data.json"

with open(origin_data_path, "r") as file:
    origin_data = json.load(file)
```

## Flatten

```python
rows = []
for item in origin_data:
    flat_item = {
        "chunk_id": item.get("chunk_id"),
        "node_type": item.get("node_type"),
        "text": item.get("text"),
        "text_len_char": len(item.get("text")),
        "text_len_word": len(item.get("text").split()),
        "has_amendment": len(item.get("amendment_notes")) > 0,
        "amendment": len(item.get("amendment_notes")),
        **item.get("metadata")
    }
    rows.append(flat_item)

df = pd.DataFrame(rows)
```

## Overall

```python
print("OVERALL--------------------")
print(f"Tổng số chunks: {len(df)}")
print(f"Số chunk_id duy nhất: {df['chunk_id'].nunique()}")
print(f"Số document_id duy nhất: {df['document_id'].nunique()}")

print("\nMISSING VALUES (%)---------")
print((round(df.isnull().sum() / len(df) * 100, 1)).sort_values(ascending=False))
```
